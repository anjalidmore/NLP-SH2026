```python
import spacy
nlp = spacy.load("en_core_web_sm")
```

```python
text = "The boys are running quickly."
doc = nlp(text)
```

```python
print("STAGE 1 — Preprocessing (Tokenization)")
for token in doc:
    print(" ", token.text)
```

```python
print("\nSTAGE 2 — Morphological (root form)")
for token in doc:
    print(f"  {token.text:12} → lemma: {token.lemma_}")
```

```python
print("\nSTAGE 3 — Syntactic (POS + dependency structure)")
for token in doc:
    print(f"  {token.text:12} POS: {token.pos_:6} depends on: {token.head.text}")
```
