```python
import spacy
nlp = spacy.load('en_core_web_sm')
sents = ["I went to the bank to deposit money.",
         "I sat by the river bank fishing."]
```

```python
for sent in sents:
    doc = nlp(sent)
    for token in doc:
        if token.text.lower() == 'bank':
            print(sent[:30], '→ POS:', token.pos_,
                  '| Entity:', token.ent_type_ or 'None')
```
