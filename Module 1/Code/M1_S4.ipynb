```python
text = "Elon Musk founded SpaceX in 2002 in Hawthorne, California."
```

# TOOL 1: NLTK

```python
import nltk; from nltk.tokenize import word_tokenize
nltk.download('punkt_tab')
nltk.download('averaged_perceptron_tagger_eng')
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')
tokens = word_tokenize(text)
ne_tree = nltk.ne_chunk(nltk.pos_tag(tokens))
for chunk in ne_tree:
    if hasattr(chunk, "label"): print("NLTK →", chunk)
```

# TOOL 2: SpaCy

```python
import spacy; nlp = spacy.load("en_core_web_sm")
for ent in nlp(text).ents: print("SpaCy →", ent.text, ent.label_)
```

# TOOL 3: HuggingFace Transformers

```python
from transformers import pipeline as hf_pipeline
ner = hf_pipeline("ner", aggregation_strategy="simple")
for r in ner(text):
    print(f"HF → {r['word']:20} {r['entity_group']} {r['score']:.3f}")
```
