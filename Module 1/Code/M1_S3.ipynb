# ---------- HuggingFace: transformer tokenization ----------

```python
text = "Hello, how are you today? I hope you are doing well."
```

```python
from transformers import AutoTokenizer
hf_tokenizer = AutoTokenizer.from_pretrained("bert-base-uncased")
hf_tokens = hf_tokenizer.tokenize(text)
```

```python
print("\nHUGGINGFACE — STAGE 1: Preprocessing (Subword Tokenization)")
for tok in hf_tokens:
    print(" ", tok)
```
