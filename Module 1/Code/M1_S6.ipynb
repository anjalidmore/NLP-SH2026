```python
import nltk, spacy
```

```python
from transformers import pipeline as hf_pipeline
```

```python
text = "The banks near the river are closed today."
```

# Stage 1: Preprocessing — Tokenization

```python
nltk.download('punkt_tab')
tokens = nltk.word_tokenize(text)
```

```python
print('Tokens:', tokens)
```

# Stage 3: Syntactic — POS Tagging + Dependencies

```python
nlp = spacy.load('en_core_web_sm')
```

```python
doc = nlp(text)
```

```python
for token in doc:
    print(token.text, token.pos_, token.dep_)
```

# Stage 6: Application — Sentiment

```python
sentiment = hf_pipeline('sentiment-analysis')
```

```python
print(sentiment(text))
```
